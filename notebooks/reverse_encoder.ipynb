```python

import numpy as np
import os
import glob
from google.colab import drive
import time


INPUT_DIR = '/content/drive/MyDrive/DNA_Anomaly_Detection/normal_batches/'
OUTPUT_DIR = '/content/drive/MyDrive/DNA_Anomaly_Detection/raw_normal_batches/'

os.makedirs(OUTPUT_DIR, exist_ok=True)

# 2. Çeviri Sözlüğü (0'lar silinir, 1-5 harfe döner)
num_to_char = np.array(['', 'A', 'C', 'G', 'T', 'N'])

def decode_batch_fast(batch_array):
    """
    Sayıları inanılmaz hızlı bir şekilde DNA harflerine çevirir.
    """
    char_array = num_to_char[batch_array]
    # Her satırı birleştirip bir liste yapar
    return ["".join(row) for row in char_array]

# 3. Dosyaları Bul
batch_files = glob.glob(os.path.join(INPUT_DIR, '*.npy'))
total_files = len(batch_files)

print(f"Toplam {total_files} dosya bulundu. Dönüşüm başlıyor...")
start_time = time.time()

# 4. İşlem Döngüsü
for i, filepath in enumerate(batch_files):
    filename = os.path.basename(filepath)
    new_filename = filename.replace('.npy', '.txt')
    output_path = os.path.join(OUTPUT_DIR, new_filename)

    # Çökme durumunda kaldığı yerden devam etmesi için kontrol
    if os.path.exists(output_path):
        continue

    try:
        int_data = np.load(filepath)

        # Yazılara çevir
        string_data = decode_batch_fast(int_data)

        # Text dosyasına tek seferde, hızlıca yaz!
        with open(output_path, 'w') as f:
            f.write('\n'.join(string_data))

    except Exception as e:
        print(f"{filename} dosyasında HATA: {e}")

    # Her 5 dosyada bir bilgi ver
    if (i + 1) % 5 == 0 or (i + 1) == total_files:
        elapsed = time.time() - start_time
        print(f"İşlenen: {i + 1}/{total_files} dosya. Geçen süre: {elapsed/60:.2f} dakika.")

print("İşlem Tamamlandı! Bu veriler DNABERT-2 için hazır.")
```
